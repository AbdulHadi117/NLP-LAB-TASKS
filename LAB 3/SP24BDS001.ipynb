### **TASK 1**

```python
import re
text = "Ali bought 25 apples in 2024.The total cost was $150, and Ahmed bought 12 more."
```

```python
# a) Extract all digits
digits = re.findall(r'\d', text)
print("Digits:", digits)
```

```python
# b) Words that start with a capital letter
capital_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text)
print("Capitalized words:", capital_words)
```

```python
# c) Sequences of exactly 4 digits
four_digits = re.findall(r'\b\d{4}\b', text)
print("4-digit sequences:", four_digits)
```

```python
# d) Words between 3 to 6 characters
words_3_to_6 = re.findall(r'\b[A-Za-z]{3,6}\b', text)
print("3-6 character words:", words_3_to_6)
```

### **TASK 2**

```python
import re

text = """
Contact ali@gmail.com or sara123@yahoo.com.
Call us at 0300-1234567 or 0312-7654321.
Order ID ORD-2024-5678 was placed on 25/09/2026.
The discount is 45% and the tax is 12.5%.
"""
```

```python
# a) Email addresses
emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)
print("Emails:", emails)
```

```python
# b) Phone numbers
phones = re.findall(r'\b03\d{2}-\d{7}\b', text)
print("Phone numbers:", phones)
```

```python
# c) Dates
dates = re.findall(r'\b\d{2}/\d{2}/\d{4}\b', text)
print("Dates:", dates)
```

```python
# d) Percentages
percentages = re.findall(r'\b\d+(?:\.\d+)?%', text)
print("Percentages:", percentages)
```

### **TASK 3**

```python
import nltk
import spacy

nltk.download("punkt")
nltk.download("punkt_tab")

text = """
I don't like waiting for buses.
It's usually very crowded in the morning.
My brother doesn't drive because he prefers cycling.
We can't always arrive on time.
"""
```

```python
# NLTK
nltk_tokens = nltk.word_tokenize(text)

# spaCy blank pipeline
nlp = spacy.blank("en")
doc = nlp(text)
spacy_tokens = [token.text for token in doc]

print("NLTK tokens:")
print(nltk_tokens)

print("\nspaCy tokens:")
print(spacy_tokens)
```

```python
# Differences:
# 1. Both separate contractions, but tokenization rules can differ for punctuation.
# 2. NLTK and spaCy may produce different token boundaries for contractions and punctuation.

print("\nNLTK token count:", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
```

### **TASK 4**

```python
import spacy

nlp = spacy.blank("en")

text = "Ali bought 5 books for $50 and emailed them to ali@example.com."
```

```python
doc = nlp(text)

for token in doc:
    print(
        token.text,
        "| is_alpha:", token.is_alpha,
        "| like_num:", token.like_num,
        "| like_email:", token.like_email
    )
```

```python
# Currency tokens
currency_tokens = [token.text for token in doc if token.is_currency]
print("\nCurrency tokens:", currency_tokens)
```
